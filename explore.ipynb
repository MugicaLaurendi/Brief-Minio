```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
urls = [
    "source/LineA_Stable_10K.csv",
    "source/LineB_Flux.csv",
    "source/LineC_Turbulent.csv",
    "source/LineD_SpikeControl.csv",
    "source/LineE_SmoothRun.csv"
]
```

```python

for url in urls :
    df = pd.read_csv(f'{url}')
    print(f"{url}".replace('.csv','').replace('source/','').replace('_',' '))
    print("--------------------------------------------------")
    print(df.head(5))
    print("--------------------------------------------------")
    print(df.describe())

    df.plot(subplots=True, layout=(1, 4), figsize=(15, 4),style='o')

    plt.tight_layout()
    plt.show()
```
